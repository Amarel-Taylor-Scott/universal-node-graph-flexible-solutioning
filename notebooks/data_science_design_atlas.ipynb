# Data-science design atlas quickstart

This dependency-free notebook compiles an all-visible design worklist. It does not execute a model, access a network, or turn catalog entries into implementation claims.

```python
from solutiongraph.design_atlas import DesignContext, DesignPlanner
from solutiongraph.design_atlas.packs import DESIGN_QUESTION_BY_ID

context = DesignContext(
    id="context.notebook.regression",
    task_type="task.tabular-regression",
    objective="Predict renewal value at the customer renewal date.",
    modalities=("modality.tabular",),
    target_name="renewal_value",
    group_field="customer_id",
)
planner = DesignPlanner()
```

```python
for effort in ("E1", "E3", "E5", "E10"):
    plan = planner.plan(context, effort=effort, random_seed=17)
    print(effort, plan.to_dict()["summary"])
```

## Give selected structured questions to a human or authorized LLM

Every question supplies explicit branches, evidence requirements, an experiment template, and stop conditions. LLM use requires `available_modes=("llm",)` and `granted_permissions=("model.invoke",)`.

```python
plan = planner.plan(context, effort="E1", random_seed=17)
for item in plan.items:
    if item.status == "selected":
        question = DESIGN_QUESTION_BY_ID[item.question_id]
        print(f"\n{question.title}\n{question.prompt}")
        print("evidence:", question.required_evidence)
        print("branches:", [choice.id for choice in question.choices])
```
