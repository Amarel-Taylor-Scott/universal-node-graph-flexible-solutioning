# 05 — Data Science: Regression and Classification

This notebook demonstrates two nontrivial DAG programs. In each, the split output fans out to training and evaluation, and the trained model joins evaluation. Regression compares mean and OLS nodes against an RMSE oracle; classification compares majority and threshold nodes against an accuracy oracle. A coding harness can extend the same typed seams with richer feature, model, calibration, ensemble, and packaging node packs.

```python
from solutiongraph.examples import run_example

regression = run_example(
    "tabular-regression",
    route="all",
    artifact_root=".artifacts/tabular-regression",
)
classification = run_example(
    "tabular-classification",
    route="all",
    artifact_root=".artifacts/tabular-classification",
)
```
